# How is tensor is saved on memory: Strides

```python
# How is a tensor stored in memory
x = torch.tensor([[1,2,3],[4,5,6]])

# Even though the shape of the tenosr is (2,3)
# PyTorch stores the data as flat 1D block of memory:
# For ex:
# Memory position:  0  1  2  3  4  5
# Value:            1  2  3  4  5  6

# PyTorch uses meta data to interpert the block of memory as a 2x3 tensor

print(x.shape) # torch.Size([2, 3])
print(x.stride()) # (3, 1)
print(x.is_contiguous()) # True

# The stride tells us to jump that many location in memory to go to next column or row.
# Here if u jump 1 memory location u get x[0,1] = 2, then again 1 jump gives x[0,2] = 3 etc..
# 3 jumps & u go to next row: x[0,3] = 4.
#
# Move down a row → jump 3
# Move right a column → jump 1

# In general:
# x[i,j] = i * stride[0] + j * stride[1] 
# x[1,1] = 1 * 3 + 1 * 1 = 4
# Memory location 4 contains 5, so x[1,1] = 5 
print(x[1,1])
```

```python
# Thus a tensor is basically data + shape + stride

# Now, Lets take transpose of x:

y = x.t() #torch.tensor([[1,4],[2,5],[3,6]])
print(y.shape) # torch.Size([3, 2])
print(y.stride()) # (1, 3)
print(y.is_contiguous()) # False
print(y)

# y[0,0]:
# offset = 0*1 + 0*3 = 0
# memory[0] = 1
#
# y[0,1]:
# offset = 0*1 + 1*3 = 3
# memory[3] = 4
#
# y[1,0]:
# offset = 1*1 + 0*3 = 1
# memory[1] = 2
#
# y[1,1]:
# offset = 1*1 + 1*3 = 4
# memory[4] = 5
#
# y[2,0]:
# offset = 2*1 + 0*3 = 2
# memory[2] = 3
#
# y[2,1]:
# offset = 2*1 + 1*3 = 5
# memory[5] = 6


# As u will see taking transopse doesnt actually change storage, only shape and stride is changed.
# since x and y share the SAME memory:

y[0,1] = 100
# Thus 1st row , 2nd element to changes to 100, 
# but also x is affected since they share same memory:
print("\nAfter y[0,1] = 100")

print("\nx:")
print(x)

print("\ny:")
print(y)

# So y[0,1] = 0 * y_stride[0] + 1 * y_stride[1]
#            = 3
# Thus Memory location 3 is now 100.
# Position: 0   1   2   3   4   5
# Value:    1   2   3   100  4  5

# Thus now 

# So x[i,j] = i * x_stride[0] + j * x_stride[1]
#            = 3 * i + j = 3
#            => i = 0 and j = 3
#            => i = 1 and j = 0

# x[0,3] = 3 (Outside limit)
# x[1,0] = 3 = Memowry(3) = 100
# Thus x[1,0] = 100
```

# Molding a Tensor  Only changes shape

```python
# Use .view() to change the shape of a tensor to any shape with same number of total elements:
# For example, a tensor with 6 elements can be viewed
# as (2,3), (3,2), (1,6), (6,1), etc.

x = torch.arange(6)
print(x.shape)
print(x.stride())
print(x.is_contiguous())

# .view() does NOT modify x in-place.
# Since the result is not assigned, x remains unchanged.
x.view(2,3) 
print("x ",x)
print(x.shape)
print(x.stride())
print(x.is_contiguous())



x = x.view(2,3) # Store the returned view
print("x ",x)
print(x.shape)
print(x.stride())
print(x.is_contiguous())

# Different valid views of the same 6 elements
x = x.view(3,2)
print("x ",x)

x = x.view(1,6)
print("x ",x)

x = x.view(6,1)
print("x ",x)


# x = x.view(3,3)
# print("x ",x)
# Throws error since 3x3 = 9 doesnt match total number of elements 6.

# Original tensor contains 6 elements
# Requested shape requires 3×3 = 9 elements
#
# Number of elements must always be preserved

# Important:
# .view() only changes how we interpret the data.
# It does NOT reorder or copy the underlying memory.
#
# Therefore .view() is very fast.
#
# Since the memory is shared, modifying the view
# will also modify the original tensor.
```

# Indexing and Slicing

```python
# Indexing selects specific elements, rows, columns, or blocks.
# ':' means "select all elements along that axis".
# Integer indexing removes the selected axis:
#     x[:,1]   -> shape (3,)
# Slicing preserves the axis:
#     x[:,1:2] -> shape (3,1)
# Rule: integer index removes an axis, slice preserves it.
```

```python
x = torch.arange(12).view(2,2,3) # Example of a  3D tensor
print("x", x)
print(x.stride())
```

```python
print(x[:,1,:]) # All blocks, row 1, all elements
print(x[:,1,:].shape) # (2,3)

# Original shape: (2,2,3)
#                 ^ ^ ^
#              block row element
#
# We fixed the row axis using integer indexing (1),
# so that axis is removed.
#
# Remaining axes:
#   block   -> size 2
#   element -> size 3
#
# Resulting shape: (2,3)

# Note here we get shape (2,3) since selecting 2nd row from both the block we loose that axis. 
# so the shape is 2 for blocks and 3 for elements
```

```python
print(x[:,:,0])
print(x[:,:,0].shape)

# Here again we select 1st element from all rows in each block. But we loose that dimension/axis since we use integer slicing
# so size is (2,2)
# To preserve that dimension:
print(x[:,:,0:1])
print(x[:,:,0:1].shape)
# Shape = (2,2,1)
#
# Here 0:1 is a slice, not an integer index.
# Slicing preserves the axis, so the element axis
# remains with size 1.
```

```python
# Element wise multiplication
#   [a,b,c] * [x,y,z]
# = [a*x, b*y, c*z]
#
# Shapes must be compatible.
# If shapes differ, PyTorch tries BROADCASTING.
#
# Broadcasting involves replicating smaller arrays to match the dimensions of larger ones during arithmetic operations.
# Rules: start from right, take first shape of both tensors, if same or if any is 1 then works else fails.
#       then check next shape towads left. If fewer dimension, adds 1. 
# (m,n) * (m,n) => (m,n)
# (m,n) * (n,) => right-align => (m,n) * (1,n) => works = (m,n)
# (m,n) * (m,) => right-align => (m,n) * (1,m) => fails (n vs m)
# (n,) * (m,n) => right-align => (1,n) * (m,n) => works = (m,n)
# (m,) * (m,n) => right-align =>  (1,m) * (m,n) =>failes (m vs n)
# (m,n) * (m,1) => works = (m,n) row-wise broadcast
# (m,n) * (1,n) => works = (m,n) column wise broadcast

matrix1 = torch.arange(6).view(2,3)
matrix2 = torch.arange(3)
result = matrix1 * matrix2
result.shape

# These rules apply to any arithmetic operations like  addition, subtraction, division, comparisons, etc.
```

```python
# -----------------------------------------------------------------------------
# HIGHER DIMENSIONS
# -----------------------------------------------------------------------------
#
# (5,2,3) * (3,)
#
# Right-align:
#
# (5,2,3)
# (1,1,3)
#
# Compare:
#
# 3 vs 3 -> OK
# 2 vs 1 -> OK
# 5 vs 1 -> OK
#
# Result: (5,2,3)
#
# -----------------------------------------------------------------------------
#
# (4,5,2,3) * (5,1,3)
#
# Right-align:
#
# (4,5,2,3)
# (1,5,1,3)
#
# Compare:
#
# 3 vs 3 -> OK
# 2 vs 1 -> OK
# 5 vs 5 -> OK
# 4 vs 1 -> OK
#
# Result: (4,5,2,3)
#
# -----------------------------------------------------------------------------
# Memory Trick
# -----------------------------------------------------------------------------
#
# Always:
#
#   1. Right-align
#   2. Pad missing dimensions with 1s on the LEFT
#   3. Compare from RIGHT to LEFT
#   4. Equal or 1 -> OK
#   5. Otherwise -> ERROR
#
# =============================================================================
```

```python
# NN use broadcsting for bias:
n_samples = 10
n_features = 3
feature_matrix = torch.arange(n_samples*n_features).view(n_samples,n_features) # (10,3)
# 1 hidden layer with 8 neurons
weights = torch.randn(n_features,8)
bias = torch.randn(8)
z = feature_matrix @ weights + bias # (10,3) @ (3,8) + (8) => (10,8) + (8) => right-align => (10,8) + (1,8) => bias broadcast(10,8)
# bias = [b1,b2,b3,b4,b5,b6,b7,b8]
# broadcast
# # bias = [
# [b1,b2,b3,b4,b5,b6,b7,b8], #1
# [b1,b2,b3,b4,b5,b6,b7,b8],
# [b1,b2,b3,b4,b5,b6,b7,b8],
# [b1,b2,b3,b4,b5,b6,b7,b8],
# [b1,b2,b3,b4,b5,b6,b7,b8],
# [b1,b2,b3,b4,b5,b6,b7,b8],
# [b1,b2,b3,b4,b5,b6,b7,b8],
# [b1,b2,b3,b4,b5,b6,b7,b8],
# [b1,b2,b3,b4,b5,b6,b7,b8],
# [b1,b2,b3,b4,b5,b6,b7,b8] # 10
# ]

# X : (10,3)
# W : (3,8)
# b : (8,)
# Z : (10,8)
```

# Dynamic Computation Graph

```python
# One of the main reasons for using PyTorch in deep learning is its
# automatic differentiation (autograd) system.

# During training, we usually define a loss/cost function and try to
# optimize the model parameters (weights and biases) by minimizing
# this loss.

# To update the parameters, we need to know how the loss changes
# with respect to small changes in each parameter, i.e. we need
# partial derivatives (gradients).

# A loss function is typically built from many operations:
# matrix multiplications, additions, activations, etc.

# As these operations are executed, PyTorch records them and builds
# a dynamic computation graph describing how every tensor was produced.

# During the backward pass, PyTorch applies the chain rule through
# this graph to automatically compute gradients.

# We only need to tell PyTorch which tensors should track gradients,
# typically the model parameters (weights and biases).

# This is done by setting:
#
#     requires_grad=True
#
# when creating the tensor.
```

```python
# In order to get familiar with the concept of a computation graph, we will create one for the following function:
```

$$
y = \frac{1}{l(x)} \sum_{i} [ (x_{i}+2)^{2} + 3 ]
$$

```python
# Lets first compute it ourselves:
# x = [0,1,2]

# a = x + 2
#  = [2,3,4]

# b = a^2
#  = [4,9,16]

# c = b + 3
#  = [7,12,19]

# y = mean(c)
#  = (7 + 12 + 19) / 3
#  = 38 / 3
#  = 12.6667
```

```python
# Here we assume that x 's is our parametes. l(x) is its size
x = torch.arange(3,dtype=torch.float32,requires_grad=True)
print(x)
```

```python
a = x + 2 
print(a.grad_fn)
print(a.grad_fn.next_functions) # no operations on x except addition with 2, so None

b = a ** 2
print(b.grad_fn)
print(b.grad_fn.next_functions)

c = b + 3
print(c.grad_fn)

y = c.mean()
print(y.grad_fn)
```

```python
y.backward()
```

```python
print(x.grad)
```

```python
def walk(fn):
    if fn is None:
        return
    print(type(fn).__name__)

    for grad, _ in fn.next_functions:
        walk(grad)


walk(y.grad_fn)
```

```python
# Even if u combine two operations, internally pytorch creates additionally nodes for each individual operations.
# i.e. PyTorch builds the graph from operations..
a = (x + 2) ** 2
print(a.grad_fn) # PowBackward
print(a.grad_fn.next_functions) # AddBackward

c = b + 3
print(c.grad_fn)
y = c.mean()
print(y.grad_fn)
```
