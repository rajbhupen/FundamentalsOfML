# Cross-Entropy Loss

In binary classification, the true label belongs to one of two classes:

$$
y_{\text{true}} \in \{0,1\}
$$

The model predicts a probability

$$
\hat y \in (0,1)
$$

using the sigmoid activation function

$$
\hat y = \sigma(z)=\frac{1}{1+e^{-z}}
$$

The binary cross-entropy loss is defined as

$$
J = -\frac{1}{N}\sum_{i=1}^{N}
\left[
y_i \log(\hat y_i)
+
(1-y_i)\log(1-\hat y_i)
\right]
$$


## Why Cross-Entropy Works

The loss becomes very small when the prediction is correct and confident:
- If $$y_i=1$$ (true class = signal) and $$\hat y_i \approx 1$$ (predicted as signal), or if $$y_i=0$$ (true class = background) and $$\hat y_i \approx 0$$ (predicted as background), the loss is very small.

- However, when the prediction is confidently wrong, i.e. $$y_i=1,\ \hat y_i \to 0$$ or $$y_i=0,\ \hat y_i \to 1$$, the loss becomes very large.

This happens because

$$
\log(0)\to -\infty
$$

Thus, cross-entropy heavily penalizes incorrect predictions.

Note that terms such as

$$
x\log x
$$

remain well-defined near zero since

$$
\lim_{x\to0^+} x\log x = 0
$$


```python

# Calculate logistic regression loss (binary corss-entropy)
def sigmoid(z):
    z = np.clip(z, -500, 500)
    s = 1 / (1 + np.exp(-z))
    return s

def calculate_logistic_loss(weights, bias, feature_matrix, labels):
    
    
    ##z = feature_matrix @ weights + bias
    z = np.dot(feature_matrix, weights) + bias
    y_pred = sigmoid(z)
   
    loss = -np.mean(labels * np.log(y_pred) + (1 - labels) * (np.log(1 - y_pred)))

    return loss
    
def gradient_descent_logistic(feature_matrix, labels, learning_rate = 0.05, n_iterations=20): # feature matrix is already transposed

    # Initialize parameters
    weights = np.zeros(feature_matrix.shape[1])  # array(n_features,)
    bias = 0.0 # scalar
    
    # Store path (Using list since it can be appended)
    w_path = [weights]
    b_path = [bias]
    loss_path = [calculate_logistic_loss(weights, bias, feature_matrix, labels)]

    
    for i in range(n_iterations):
        
        z =  np.dot(feature_matrix, weights) + bias #feature_matrix @ weights + bias
        y_pred = sigmoid(z)
    

        # Calculate gradients [Using sigmoidal function results in very simple form]
        #dw = - np.mean((labels - y_pred) @ feature_matrix)
        #dw = -(1 / len(labels)) * ((labels - y_pred) @ feature_matrix)
       # print(dw)
        #db = -np.mean(labels - y_pred)

        error = y_pred - labels
        dw = np.dot(feature_matrix.T, error) / len(labels)
        db = np.mean(error)

        
        # Update
        weights -= learning_rate * dw
        bias -= learning_rate * db

        # Append
        w_path.append(weights.copy())
        b_path.append(bias)
        loss_path.append(calculate_logistic_loss(weights, bias, feature_matrix, labels))
    
    
    return np.array(w_path), np.array(b_path), np.array(loss_path) # convert to array
        
```

```python
w_path, b_path, loss_path = gradient_descent_logistic(feature_matrix, labels, learning_rate = 0.05, n_iterations=20)
print(w_path)

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

feat1 = 0
feat2 = 1

fig, ax1 = plt.subplots(figsize=(7,6))

mask = labels == 0

feat1_data = feature_matrix[:, feat1]
feat2_data = feature_matrix[:, feat2]

ax1.scatter(feat1_data[mask],
            feat2_data[mask],
            color='red',
            label='Apartments')

ax1.scatter(feat1_data[~mask],
            feat2_data[~mask],
            color='blue',
            label='House')

ax1.set_xlabel('Feature 1')
ax1.set_ylabel('Feature 2')

# Meshgrid
x_vals = np.linspace(feat1_data.min(), feat1_data.max(), 100)
y_vals = np.linspace(feat2_data.min(), feat2_data.max(), 100)

X, Y = np.meshgrid(x_vals, y_vals)

ax1.grid()
ax1.legend()

# Store contour object
contour_plot = None

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def animate(frame):

    global contour_plot

    step = min(frame, len(w_path)-1)

    current_w = w_path[step]
    current_b = b_path[step]

    # Remove previous contour
    if contour_plot is not None:
        for coll in contour_plot.collections:
            coll.remove()

    # Create full feature matrix for grid
    Z = np.zeros_like(X)

    # Contribution from displayed features
    Z += current_w[feat1] * X
    Z += current_w[feat2] * Y

    # Contribution from hidden features
    for j in range(feature_matrix.shape[1]):

        if j != feat1 and j != feat2:

            Z += current_w[j] * np.mean(feature_matrix[:, j])

    Z += current_b

    P = sigmoid(Z)

    # Decision boundary at probability = 0.5
    contour_plot = ax1.contour(
        X,
        Y,
        P,
        levels=[0.5],
    )

    ax1.set_title(f"Iteration {step}")

    return contour_plot.collections

anim = FuncAnimation(
    fig,
    animate,
    frames=len(w_path),
    interval=400,
    repeat=True
)

HTML(anim.to_jshtml())
```
